# Hypothesis and Findings

## Hypothesis

There is a relationship between menstrual cycle phases and physical and emotional states such as mood, cramps, and energy levels.

## Findings

The analysis reveals clear and consistent patterns across different phases of the menstrual cycle.

Mood is highest during the non-menstrual (other) phase and drops significantly during the premenstrual phase. This indicates that emotional well-being is negatively affected as the cycle approaches menstruation.

Cramps are highest during the menstrual phase, while they are nearly absent during the rest of the cycle. A slight increase is observed in the premenstrual phase, suggesting early physical discomfort before menstruation begins.

Energy levels follow a similar pattern to mood. Energy is highest during the non-menstrual phase and decreases during both menstrual and premenstrual phases. The lowest energy levels are observed just before menstruation.

Overall, these findings strongly support the hypothesis that menstrual cycle phases influence both physical symptoms and emotional well-being.

## Statistical Testing (Correlation Analysis)

```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
df = pd.read_csv("clue_daily_data.csv")

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date")
```

```python
from scipy.stats import pearsonr, spearmanr

pearson_corr, _ = pearsonr(df["mood"], df["cramps"])
spearman_corr, _ = spearmanr(df["mood"], df["cramps"])

print("Pearson correlation (mood vs cramps):", pearson_corr)
print("Spearman correlation (mood vs cramps):", spearman_corr)
```

The correlation results show a negative relationship between mood and cramps.

Both Pearson and Spearman coefficients indicate that as cramps increase, mood tends to decrease.

This statistically supports the hypothesis that physical discomfort is associated with lower emotional well-being.

```python
pearson_corr_e, _ = pearsonr(df["energy"], df["cramps"])
print("Pearson (energy vs cramps):", pearson_corr_e)
```

The correlation between energy and cramps shows a negative relationship.

This suggests that as cramps increase, energy levels tend to decrease. This finding is consistent with the earlier observations in the EDA phase, where lower energy levels were associated with days of higher physical discomfort.

Overall, this result further supports the hypothesis that physical symptoms have a measurable impact on daily well-being.
